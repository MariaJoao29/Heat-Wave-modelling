Now, I want to create a function that applied the wind chill index, considering the temperature and the wind velocity

```python
def wind_chill(T_air, M):
    """
    Calculate the wind chill index.
    T_air: Air temperature in ºC
    M: Wind speed in m/s
    Returns: Wind chill temperature in ºC
    """
    M0 = 4.8  # Km/h
    M = M * (3600 / 1000)  # Convert wind speed from m/s to km/h
    
    if M <= M0 and T_air < 10:
        return T_air  # No wind chill if wind is below threshold and temp is above 10ºC
    elif M <= M0 and T_air >= 10:
        return 'NA'  # Not applicable
    elif M > M0 and T_air >= 10:
        return 'NA'  # Not applicable
    elif M > M0 and T_air < 10:
        # Wind chill formula
        return 13.12 + 0.6215 * T_air - 11.37 * M ** 0.16 + 0.3965 * T_air * M ** 0.16

# List of input CSV file paths
csv_files = [
    r'C:\Users\maria\HW_Data\Data\erp_timeseries_1996-2015_converted.csv',
    r'C:\Users\maria\HW_Data\Data\epp_timeseries_1996-2015_converted.csv',
    r'C:\Users\maria\HW_Data\Data\etp_timeseries_2081-2100_converted.csv'
]

# Process each CSV file
for csv in csv_files:
    # Read the CSV file into a DataFrame
    df = pd.read_csv(csv)
    
    # Compute wind chill for different temperature columns
    df['wind_chill'] = df.apply(lambda row: wind_chill(row['tas'], row['sfcwind']), axis=1)
   # df['wind_chill_max'] = df.apply(lambda row: wind_chill(row['tasmax'], row['sfcwind']), axis=1)
    #df['wind_chill_min'] = df.apply(lambda row: wind_chill(row['tasmin'], row['sfcwind']), axis=1)
    
    # Save the updated DataFrame to a new CSV file
    new_csv = csv.replace('_converted.csv', '_with_wind_chill.csv')
    df.to_csv(new_csv, index=False)
    

print("Wind chill calculations completed and saved to new files.")
```
